```python
import datetime
from pyspark.sql import functions as F

orders= spark.read.csv('/Volumes/workspace/default/sample_orders_csv/', inferSchema=True, header=True)
display(orders)

bronze_orders_new = orders.withColumn('ingestion_timestamp', F.current_timestamp())
display(bronze_orders_new)
(
    bronze_orders_new
    .write
    .format('delta')
    .mode('overwrite')
    .option('overwriteSchema', 'true')
    .saveAsTable('default.bronze_orders_new')
)
```
